## CSDI modèle 

Paramètres d'entrée :

- `n_steps`: int, number of time steps in the time-series data sample
- `n_feature`: int, number of the feature in the time-series data sample
- `n_layers`: int, number of layers in the CSDI model
- `n_heads`: int, number of heads in the multi-head attention mechanism
- `n_channel`: int, number of residual channels
- `d_time_embedding`: int, dimension number of the time embedding
- `d_feature_embedding` : int, dimension number of the feature embedding
- `d_diffusion_embedding`: int, dimension number of the diffusion embedding
- `is_unconditional` : bool,  weather the model is unconditional or conditional
- `n_diffusion_steps` : int, number of the diffusion step T in the original paper
- `schedule` : str, "quad" or "linear", schedule for other noise levels.
- `beta_start` : float, minimum noise level
- `beta_end` : float, maximum noise level
- `batch_size`: int, batch size for training and evaluating the model
- `epochs`: int, number of epochs for training the model 
- `patience`:[optional], int The patience for the early-stopping mechanism. Given a positive integer, the training process will be stopped when the model does not perform better after that number of epochs. Leaving it default as None will disable the early-stopping.
- `device` : "cpu" for now, gpu later maybe

# Préparation des données

On permute les données pour qu'elles aient la bonne taille pour le réseau et on met les données sous forme de dictionnaire.

## Side information 

Permet de construire l'information auxiliaire donnée au débruiteur pour chaque couche. Va savoir où on est dans le temps, savoir quelle est la variable et a le masque conditionnel pour savoir quels points sont observés

## Construction du modèle

Création du réseaux et planning du bruit

## Débruiteur

Prend un pas t au hasard, s'informe sur le niveau de bruit, corrélations temporelles et spatiales grâce à side info et estime le bruit.

## Calcul du loss

Pour chaque pas t pris dans le débruiteur, calcule le loss du bruit estimé. Calcule ensuite le loss total sur tous les pas de temps. 

## Diffusion inverse

Par d'un bruit pur et débruite de T à 0 en utlisant l'entrainement du modèle. Fais ce processus `n_sampling_times` fois pour évaluer toutes les imputations plausibles.

## Forward

Imputation selon les valeurs d'imputations qui minimisent la loss.

# Fonction globale fit() et fonctions annexes

- `fit()` déduit les masques des NaN et tire lui-même le masquage d'entraînement. On fournit nos propres masques.
- `make_cfg` : configuration par défaut du modèle 
- `masked_stats`, `normalize` : standardisation par variable, **uniquement sur les points observés** du train
- `make_windows`, `select_windows` : découpage de la longue série (≈ 105 000 heures) en fenêtres de longueur `L`, indispensable car l'attention temporelle coûte O(L²)
- `fit_arrays` : entraînement (masquage aléatoire par `get_rand_mask`, ou `cond_mask` fourni via `Ctr`)
- `impute` : imputation avec un `cond_mask` explicite (les points cachés sont mis à 0 avant d'entrer dans le modèle)
- `masked_metrics`, `per_variable_metrics`, `interp_baseline` : évaluation sur les cibles du carving et comparaison à des méthodes simples
- `sanity_check` : vérifie les formes et la loss initiale avant de lancer un vrai entraînement

# Lancement

1. **Paramètres** : `QUICK_TEST = True` entraîne un petit modèle sur un sous-ensemble de fenêtres, pour vérifier que tout tourne. Passer à `False` pour le vrai entraînement.
2. **Préparation** : standardisation sur le train, fenêtres, test avec le carving.
3. **Entraînement**, puis **évaluation** sur les points cachés par le carving, comparée à deux méthodes simples (la moyenne et l'interpolation linéaire).


```python
#  Préparation des données
names = list(df_train.columns) # noms des K variables

# On divise les données d'entrainement train et validation
X_tr_raw = df_train.to_numpy(np.float32)# (L_tot,K)
O_tr_raw = observed_mask[0].numpy()# (L_tot,K)
mean, std = masked_stats(X_tr_raw, O_tr_raw)# stats du TRAIN, sur les points observés
X_tr_n = normalize(X_tr_raw, O_tr_raw, mean, std)

#on transforme les données en fenêtres temporelles
cut = int(0.9 * len(X_tr_n)) # split temporel 90 % / 10 %
Xtr_w = make_windows(X_tr_n[:cut], L, L // 2)
Otr_w = make_windows(O_tr_raw[:cut], L, L // 2)
Xva_w = make_windows(X_tr_n[cut:], L)
Ova_w = make_windows(O_tr_raw[cut:], L)
keep_tr = Otr_w.mean(axis=(1, 2)) >= MIN_OBS                          
keep_va = Ova_w.mean(axis=(1, 2)) >= MIN_OBS

if training:   # on découpe le masque d'entrainement en fenêtres
    C_tr_raw = cond_mask_train[0].numpy()# (L_tot,K)
    Ctr_w = make_windows(C_tr_raw[:cut], L, L // 2)
    Cva_w = make_windows(C_tr_raw[cut:], L)
    keep_tr &= (Otr_w - Ctr_w).sum(axis=(1, 2)) > 0# au moins une cible par fenêtre, sinon sa loss vaut 0
    keep_va &= (Ova_w - Cva_w).sum(axis=(1, 2)) > 0
    
i_tr = select_windows(keep_tr, MAX_TR)
i_va = select_windows(keep_va, MAX_VA)
Xtr_w, Otr_w, Xva_w, Ova_w = Xtr_w[i_tr], Otr_w[i_tr], Xva_w[i_va], Ova_w[i_va]
if training:
    Ctr_w, Cva_w = Ctr_w[i_tr], Cva_w[i_va]
    print(f"cibles d'apprentissage : train {int((Otr_w - Ctr_w).sum())} | validation {int((Ova_w - Cva_w).sum())}")
    
# Test 
X_te_raw = df_test.to_numpy(np.float32)
O_te_raw = observed_mask_test[0].numpy()
C_te_raw = cond_mask[0].numpy()
X_te_n = normalize(X_te_raw, O_te_raw, mean, std)# mêmes mean/std que le train
Xte_w, Ote_w, Cte_w = (make_windows(a, L) for a in (X_te_n, O_te_raw, C_te_raw))
i_te = select_windows((Ote_w - Cte_w).sum(axis=(1, 2)) > 0, MAX_TE)# fenêtres qui contiennent des cibles
Xte_w, Ote_w, Cte_w = Xte_w[i_te], Ote_w[i_te], Cte_w[i_te]
target_te = Ote_w - Cte_w

print("train :", Xtr_w.shape, "| val :", Xva_w.shape, "| test :", Xte_w.shape)
print(f"taux d'observation train : {Otr_w.mean():.2%} | cibles dans le test : {int(target_te.sum())} ({target_te.sum() / Ote_w.sum():.1%} des points observés)")
```

## Entraînement 

```python
if training :
    model, history = fit_arrays(Xtr_w, Otr_w, Xva_w, Ova_w, Ctr=Ctr_w, Cva=Cva_w,epochs=EPOCHS, batch_size=BATCH,norm=dict(mean=mean, std=std), save_path="csdi_best.pt",device = device ,**CFG)
    pd.DataFrame(history).to_csv(path_history, index=False)
else :
    #reconstruction du modèle sans entraînement
    ckpt = torch.load(path_model, weights_only=False, map_location=device )
    history = pd.read_csv(path_history).to_dict("records")
    device = "cuda" if torch.cuda.is_available() else "cpu"

    model = build_model(ckpt["cfg"], device)     
    model["net"].load_state_dict(ckpt["net"])   
    model["net"].eval()
    mean, std = ckpt["norm"]["mean"], ckpt["norm"]["std"]
```

## Imputation

attention peut durer moins de 3h

```python
if imputation :
    imp = impute(model, Xte_w, Cte_w, n_sampling_times=N_SAMPLES)  
    
    #Sauvegarde :
    np.savez_compressed(
        "imputation_csdi.npz",
        imp=imp,# (N,L,K) imputations normalisées (médiane)
        i_te=i_te,# indices des fenêtres : la fenêtre n couvre les heures i_te[n]*L ... (i_te[n]+1)*L
        L=L, mean=mean, std=std,# pour revenir aux unités physiques
        names=np.array(names),# noms des K variables
        x_true=Xte_w, cond=Cte_w, observed=Ote_w,# vérité, contexte visible, points observés (normalisés)
        n_sampling_times=N_SAMPLES,
)
else :
    d = np.load(path_imp)
    imp, i_te, L = d["imp"], d["i_te"], int(d["L"])
    mean, std, names = d["mean"], d["std"], list(d["names"])
    Xte_w, Cte_w, Ote_w = d["x_true"], d["cond"], d["observed"]
    target_te = Ote_w - Cte_w              
    
mae, rmse = masked_metrics(imp, Xte_w, target_te)
mae0, rmse0 = masked_metrics(np.zeros_like(Xte_w), Xte_w, target_te)  #imputer la moyenne (0 en unités normalisées)
lin = interp_baseline(Xte_w, Cte_w)
mae1, rmse1 = masked_metrics(lin, Xte_w, target_te) #interpolation linéaire

print(pd.DataFrame({"MAE": [mae, mae0, mae1], "RMSE": [rmse, rmse0, rmse1]},index=["CSDI", "Moyenne", "Interpolation linéaire"]).round(3))
print()
print(per_variable_metrics(imp, Xte_w, target_te, std, names).round(3))

```

La loss correspond à l'erreur quadratique entre le bruit ajouté aux données et le bruit prédit par le modèle. Elle est calculée uniquement sur les points cachés.

La courbe bleu correspond à la loss à chaque epoch lors de l'entrainement du modèle. Pour un même epoch on test sur les fenêtres non utilisées pour l'apprentissage.

Un epoch correspond à un passage sur toutes les fenêtres temporelles.
