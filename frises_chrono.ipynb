## IMF & PLASMA

```python
import matplotlib.dates as mdates
import matplotlib.patches as mpatches
import matplotlib.pyplot as plt
import pandas as pd

# Chemin vers votre fichier Excel (à adapter si nécessaire)
file_path = "C:\\Users\\pierr\\OneDrive - IPSA\\Documents\\IPSA\\Aero 5\\PMI\\OMNI_data_availability.xlsx"

# 1. Chargement des deux feuilles spécifiques du fichier Excel
df_imf = pd.read_excel(file_path, sheet_name='LRO_IMF_data_sources')
df_plasma = pd.read_excel(file_path, sheet_name='LRO_plasma_data_sources')

# 2. Nettoyage des lignes vides pour éviter les erreurs de type
df_imf = df_imf.dropna(subset=['Spacecraft', 'date début', 'durée'])
df_plasma = df_plasma.dropna(subset=['Spacecraft', 'date début', 'durée'])

# 3. Conversion de la date de début au format datetime et calcul numérique
df_imf['date début'] = pd.to_datetime(df_imf['date début'], format='%d/%m/%Y')
df_imf['start_num'] = mdates.date2num(df_imf['date début'])

df_plasma['date début'] = pd.to_datetime(
    df_plasma['date début'], format='%d/%m/%Y'
)
df_plasma['start_num'] = mdates.date2num(df_plasma['date début'])

# 4. Tri des spacecrafts par date de début la plus précoce (combinant IMF et Plasma)
df_all_dates = pd.concat([
    df_imf[['Spacecraft', 'date début']],
    df_plasma[['Spacecraft', 'date début']],
])
earliest_dates = (
    df_all_dates.groupby('Spacecraft')['date début'].min().reset_index()
)
earliest_dates = earliest_dates.sort_values('date début', ascending=True)
all_spacecrafts = earliest_dates['Spacecraft'].tolist()

# Création d'un dictionnaire pour associer chaque spacecraft à une position Y fixe
sc_to_y = {sc: i for i, sc in enumerate(all_spacecrafts)}

# Ensembles pour vérifier la présence
imf_spacecrafts = set(df_imf['Spacecraft'])
plasma_spacecrafts = set(df_plasma['Spacecraft'])

# 5. Création du diagramme de Gantt combiné
fig, ax = plt.subplots(figsize=(14, 9))

bar_height = 0.35  # Hauteur des barres pour permettre la superposition propre

# Traçage des barres IMF (en haut de la ligne si partagé)
for _, row in df_imf.iterrows():
  sc = row['Spacecraft']
  y_pos = sc_to_y[sc]
  offset = 0.18 if sc in plasma_spacecrafts else 0.0

  ax.barh(
      y_pos + offset,
      row['durée'],
      left=row['start_num'],
      height=bar_height,
      color='darkcyan',
      edgecolor='darkslategray',
      alpha=0.85,
  )

# Traçage des barres Plasma (en bas de la ligne si partagé)
for _, row in df_plasma.iterrows():
  sc = row['Spacecraft']
  y_pos = sc_to_y[sc]
  offset = -0.18 if sc in imf_spacecrafts else 0.0

  ax.barh(
      y_pos + offset,
      row['durée'],
      left=row['start_num'],
      height=bar_height,
      color='mediumvioletred',
      edgecolor='darkslategray',
      alpha=0.85,
  )

# 6. Configuration des étiquettes de l'axe Y (uniquement le nom du spacecraft)
ax.set_yticks(list(sc_to_y.values()))
ax.set_yticklabels(all_spacecrafts)

# 7. Formatage de l'axe X (affichage des années)
ax.xaxis_date()
ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))
ax.xaxis.set_major_locator(mdates.YearLocator(5))

# 8. Ajout de la légende
imf_patch = mpatches.Patch(color='darkcyan', alpha=0.85, label='IMF')
plasma_patch = mpatches.Patch(
    color='mediumvioletred', alpha=0.85, label='Plasma'
)
ax.legend(handles=[imf_patch, plasma_patch], loc='upper right', fontsize=11)

# 9. Mise en page et esthétique
ax.set_xlabel('Années', fontsize=12, fontweight='bold')
ax.set_ylabel('Spacecraft (ID)', fontsize=12, fontweight='bold')
ax.set_title(
    'Disponibilité combinée des sources IMF et Plasma',
    fontsize=14,
    fontweight='bold',
)

# Inverser l'axe Y pour afficher le premier vaisseau en haut
plt.gca().invert_yaxis()

plt.grid(axis='x', linestyle='--', alpha=0.6)

ax.set_xlim(right=pd.to_datetime('2026-12-31'))
plt.tight_layout()

# Sauvegarde optionnelle et affichage
# plt.savefig('combined_gantt.png', dpi=300)
plt.show()
```

## + Prioratisation 

```python
import matplotlib.dates as mdates
import matplotlib.patches as mpatches
import matplotlib.pyplot as plt
import pandas as pd

# Chemin vers votre fichier Excel (à adapter si nécessaire)
file_path = (
    "C:\\Users\\pierr\\OneDrive - IPSA\\Documents\\IPSA\\Aero"
    " 5\\PMI\\OMNI_data_availability.xlsx"
)

# 1. Chargement des feuilles spécifiques du fichier Excel
df_imf = pd.read_excel(file_path, sheet_name='LRO_IMF_data_sources')
df_plasma = pd.read_excel(file_path, sheet_name='LRO_plasma_data_sources')
df_priorities = pd.read_excel(file_path, sheet_name='LRO_priorities')

# 2. Nettoyage des lignes vides pour éviter les erreurs de type
df_imf = df_imf.dropna(subset=['Spacecraft', 'date début', 'durée'])
df_plasma = df_plasma.dropna(subset=['Spacecraft', 'date début', 'durée'])
df_priorities = df_priorities.dropna(subset=['Debut', 'Fin'])

# 3. Conversion de la date de début au format datetime et calcul numérique
df_imf['date début'] = pd.to_datetime(df_imf['date début'], format='%d/%m/%Y')
df_imf['start_num'] = mdates.date2num(df_imf['date début'])

df_plasma['date début'] = pd.to_datetime(
    df_plasma['date début'], format='%d/%m/%Y'
)
df_plasma['start_num'] = mdates.date2num(df_plasma['date début'])

df_priorities['Debut'] = pd.to_datetime(
    df_priorities['Debut'], format='%d/%m/%Y'
)
df_priorities['Fin'] = pd.to_datetime(df_priorities['Fin'], format='%d/%m/%Y')

# 4. Tri des spacecrafts par date de début la plus précoce (combinant IMF et Plasma)
df_all_dates = pd.concat([
    df_imf[['Spacecraft', 'date début']],
    df_plasma[['Spacecraft', 'date début']],
])
earliest_dates = (
    df_all_dates.groupby('Spacecraft')['date début'].min().reset_index()
)
earliest_dates = earliest_dates.sort_values('date début', ascending=True)
all_spacecrafts = earliest_dates['Spacecraft'].tolist()

# Création d'un dictionnaire pour associer chaque spacecraft à une position Y fixe
sc_to_y = {sc: i for i, sc in enumerate(all_spacecrafts)}

# Ensembles pour vérifier la présence
imf_spacecrafts = set(df_imf['Spacecraft'])
plasma_spacecrafts = set(df_plasma['Spacecraft'])


# Fonction de correspondance robuste pour les noms de satellites
def match_spacecraft(name_str, all_scs):
  if pd.isna(name_str):
    return None
  name_str = str(name_str).strip()
  if name_str in all_scs:
    return name_str
  base = name_str.split('(')[0].strip().lower()
  for sc in all_scs:
    sc_base = sc.split('(')[0].strip().lower()
    if base == sc_base or base in sc.lower():
      return sc
  return None


# 5. Création du diagramme de Gantt combiné
fig, ax = plt.subplots(figsize=(14, 9))

bar_height = 0.35

# Traçage des barres IMF (en haut de la ligne si partagé)
for _, row in df_imf.iterrows():
  sc = row['Spacecraft']
  y_pos = sc_to_y[sc]
  offset = 0.18 if sc in plasma_spacecrafts else 0.0

  ax.barh(
      y_pos + offset,
      row['durée'],
      left=row['start_num'],
      height=bar_height,
      color='darkcyan',
      edgecolor='darkslategray',
      alpha=0.85,
  )

# Traçage des barres Plasma (en bas de la ligne si partagé)
for _, row in df_plasma.iterrows():
  sc = row['Spacecraft']
  y_pos = sc_to_y[sc]
  offset = -0.18 if sc in imf_spacecrafts else 0.0

  ax.barh(
      y_pos + offset,
      row['durée'],
      left=row['start_num'],
      height=bar_height,
      color='mediumvioletred',
      edgecolor='darkslategray',
      alpha=0.85,
  )

# 6. Ajout des frontières (thresholds) et des numéros de priorité
priority_cols = [
    col for col in df_priorities.columns if str(col).startswith('N')
]

# Ensemble pour éviter de tracer plusieurs fois la même ligne verticale (début/fin identiques)
plotted_dates = set()

for _, row in df_priorities.iterrows():
  start_num = mdates.date2num(row['Debut'])
  end_num = mdates.date2num(row['Fin'])
  mid_num = start_num + (end_num - start_num) / 2

  # Tracé des frontières (début et fin) pour bien distinguer les intervalles
  if start_num not in plotted_dates:
    ax.axvline(
        x=start_num, color='dimgray', linestyle='--', linewidth=0.8, alpha=0.7
    )
    plotted_dates.add(start_num)
  if end_num not in plotted_dates:
    ax.axvline(
        x=end_num, color='dimgray', linestyle='--', linewidth=0.8, alpha=0.7
    )
    plotted_dates.add(end_num)

  for col in priority_cols:
    val = row[col]
    if pd.notna(val):
      sc = match_spacecraft(val, all_spacecrafts)

      if sc and sc in sc_to_y:
        y_pos = sc_to_y[sc]
        offset = 0.18 if sc in plasma_spacecrafts else 0.0
        y_text = y_pos + offset - (bar_height / 2) - 0.08

        # On enlève le "N" pour n'afficher que le chiffre (ex: "N1" -> "1")
        display_text = str(col).replace('N', '')

        ax.text(
            mid_num,
            y_text,
            display_text,
            ha='center',
            va='bottom',
            fontsize=9,
            fontweight='bold',
            color='black',  # Texte en noir
            bbox=dict(
                boxstyle='round,pad=0.15',
                facecolor='white',
                edgecolor='black',
                alpha=0.8,
            ),
        )

# 7. Configuration des étiquettes de l'axe Y
ax.set_yticks(list(sc_to_y.values()))
ax.set_yticklabels(all_spacecrafts)

# 8. Formatage de l'axe X
ax.xaxis_date()
ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))
ax.xaxis.set_major_locator(mdates.YearLocator(5))

# 9. Ajout de la légende
imf_patch = mpatches.Patch(color='darkcyan', alpha=0.85, label='IMF')
plasma_patch = mpatches.Patch(
    color='mediumvioletred', alpha=0.85, label='Plasma'
)
ax.legend(handles=[imf_patch, plasma_patch], loc='upper right', fontsize=11)

# 10. Mise en page et esthétique
ax.set_xlabel('Années', fontsize=12, fontweight='bold')
ax.set_ylabel('Spacecraft (ID)', fontsize=12, fontweight='bold')
ax.set_title(
    'Disponibilité combinée des sources IMF, Plasma et Priorités',
    fontsize=14,
    fontweight='bold',
)

# Inverser l'axe Y pour afficher le premier vaisseau en haut
plt.gca().invert_yaxis()

plt.grid(axis='x', linestyle=':', alpha=0.4)

ax.set_xlim(right=pd.to_datetime('2026-12-31'))
plt.tight_layout()

# plt.savefig('combined_gantt_priorities.png', dpi=300)
plt.show()
```
